## Objective
The objective of this notebook is to represent the future
cash flows of a fixed-rate coupon bond.

A fixed-rate bond generates two types of cash flows:

1. Periodic coupon payments
2. Principal repayment at maturity

## Coupon Payment

For a fixed-rate bond,

\[
C = \frac{F \times c}{m}
\]

where:

- \(F\) = face value
- \(c\) = annual coupon rate
- \(m\) = number of coupon payments per year
